```python
plt.figure(figsize=(10,5))
sns.set_style('whitegrid')
sns.boxplot(x='day', y='total_bill', data=tips_df, hue='sex')

plt.legend(loc=0)
```

```python
sns.set_style('darkgrid')
plt.figure(figsize=(10,5))
sns.violinplot(x='day', y='tip', data=tips_df, hue='sex')
```

```python
sns.stripplot(x='day', y='tip', data=tips_df, jitter=True, hue='sex', dodge=True)
```

```python
sns.violinplot(x='day', y='tip', data=tips_df, hue='sex')
sns.swarmplot(x='day', y='tip', data=tips_df, color='white', hue='sex',dodge=True)
```

```python
plt.figure(figsize=(10,5))
sns.set_style('dark')
sns.set_context('talk')

sns.stripplot(x='day', y='tip', data=tips_df, hue='sex', palette='viridis', jitter=True, dodge=True)
```

```python
plt.figure(figsize=(8,6))
sns.set_context('paper', font_scale=1.4)
df = crash_df.pop('abbrev')

crash_mx = crash_df.corr()
sns.heatmap(crash_mx, annot=True, cmap='Blues')
```

```python
plt.figure(figsize=(10,6))
sns.set_context('paper', font_scale=1.4)
flights = sns.load_dataset('flights')

flights = flights.pivot_table(index='month', columns='year', values='passengers')
sns.heatmap(flights, cmap='Reds', linecolor='white', linewidths=1)
```

```python
plt.figure(figsize=(10,6))
sns.set_context('paper', font_scale=1.4)
iris = sns.load_dataset('iris')

#species = iris.pop("species")
#sns.clustermap(iris, cmap='Blues', standard_scale=1)
sns.clustermap(flights, cmap='Blues', standard_scale=1)
```

```python
plt.figure(figsize=(10,6))
sns.set_context('paper', font_scale=1.4)
iris_g = sns.PairGrid(iris, hue='species',
                      x_vars=['sepal_length', 'sepal_width'],
                      y_vars=['petal_length', 'petal_width'])

iris_g.map(plt.scatter, alpha=0.8)
iris_g.add_legend()
```

```python
att_df = sns.load_dataset('attention')
att_fg = sns.FacetGrid(att_df, col='subject', col_wrap=5, height=1.5)
att_fg.map(plt.plot, 'solutions', 'score', marker='.')
```

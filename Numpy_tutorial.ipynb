```python
import numpy_financial as npf
npf.fv(.05/12, 12, -0, -100) # compond interst
period = np.arange(12)+1

principle = 10000.00
ipmt =npf.ipmt(.1/12, period, 12, principle)
ppmt =npf.ppmt(.1/12, period, 12, principle)

for payment in period:
  index = payment-1
  principle = principle + ppmt[index]

  print(f'{payment} {np.round(ppmt[index], 2)} {np.round(ipmt[index], 2)} {np.round(principle, 2)}')

np.round(npf.nper(.1/12, -150, 10000), 2)

npf.npv(0.50, [-15000, 4000, 5000, 6000, 7000]).round(2)
```
